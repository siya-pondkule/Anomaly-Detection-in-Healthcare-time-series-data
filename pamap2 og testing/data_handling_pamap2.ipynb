```python
import os
import pandas as pd
import glob

base_path = r"C:\Users\aishw\Downloads\pamap2+physical+activity+monitoring\PAMAP2_Dataset\PAMAP2_Dataset\Protocol"
output_path = os.path.join(base_path, "csv")
os.makedirs(output_path, exist_ok=True)



imu_features = [
    "temperature",
    "acc16_x", "acc16_y", "acc16_z",
    "acc6_x", "acc6_y", "acc6_z",
    "gyro_x", "gyro_y", "gyro_z",
    "mag_x", "mag_y", "mag_z",
    "ori_w", "ori_x", "ori_y", "ori_z"
]

columns = ["timestamp", "activityID", "heartRate"] + \
          [f"hand_{feat}" for feat in imu_features] + \
          [f"chest_{feat}" for feat in imu_features] + \
          [f"ankle_{feat}" for feat in imu_features]

assert len(columns) == 54, f"Expected 54 columns, got {len(columns)}"

# ===============================
# 🔄 3️⃣ Convert each .dat → .csv
# ===============================
csv_files = []

for file in os.listdir(base_path):
    if file.endswith(".dat"):
        dat_file_path = os.path.join(base_path, file)
        csv_file_path = os.path.join(output_path, file.replace(".dat", ".csv"))
        print(f"Converting {file} ...")

        # Read space-separated file
        df = pd.read_csv(dat_file_path, sep=r"\s+", header=None, names=columns)

        # Drop rows with all NaNs (if any)
        df.dropna(how="all", inplace=True)

        # Optional: take only a part of data from each subject to keep file lighter
        # e.g., first 10,000 rows
        df = df.head(10000)

        # Save as CSV
        df.to_csv(csv_file_path, index=False)
        csv_files.append(csv_file_path)
        print(f"✅ Saved: {csv_file_path} ({df.shape[0]} rows)")

# ===============================
# 📊 4️⃣ Merge all CSVs into one
# ===============================
print("\nMerging all subject CSVs into pamap2_combined.csv ...")
all_dfs = [pd.read_csv(f) for f in csv_files]
combined_df = pd.concat(all_dfs, ignore_index=True)

combined_csv_path = os.path.join(output_path, "pamap2_combined.csv")
combined_df.to_csv(combined_csv_path, index=False)

print(f"🎉 Done! Combined CSV saved at:\n{combined_csv_path}")
print(f"Shape of merged dataset: {combined_df.shape}")
```
