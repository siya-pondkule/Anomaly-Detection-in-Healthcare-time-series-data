```python
#plotting

import pandas as pd
import os

base_path = r"C:\Users\aishw\.cache\kagglehub\datasets\wangboluo\pamap2\versions\1"

# Adjust file names if different
train_file = os.path.join(base_path, "train.csv")
test_file  = os.path.join(base_path, "test.csv")

train_df = pd.read_csv(train_file)
test_df  = pd.read_csv(test_file)

print("Train shape:", train_df.shape)
print("Test  shape:", test_df.shape)

print("\nTrain columns:\n", train_df.columns[:10], "...")
print("\nTrain sample:\n", train_df.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Missing values
print("\nMissing values (top 10):")
print(train_df.isna().sum().sort_values(ascending=False).head(10))

# Activity distribution
plt.figure(figsize=(12,4))
train_df['Activity'].value_counts().sort_index().plot(kind='bar')
plt.title("Activity distribution in TRAIN")
plt.xlabel("Activity ID")
plt.ylabel("Samples")
plt.show()

# Correlation among accelerometer features
acc_cols = [c for c in train_df.columns if 'acc' in c]
corr = train_df[acc_cols].corr()
plt.figure(figsize=(10,8))
sns.heatmap(corr, cmap='coolwarm', annot=False)
plt.title("Correlation among accelerometer axes (Train)")
plt.show()
```

```python
# Pick first subject or id
subject_id = train_df['id'].unique()[0]
train_sub = train_df[train_df['id']==subject_id].reset_index(drop=True)

sensor_col = 'imu_hand_accX_16g'
plt.figure(figsize=(14,4))
plt.plot(train_sub[sensor_col].iloc[:2000])
plt.title(f"{sensor_col} — Subject {subject_id} (first 2000 samples)")
plt.xlabel("Sample index")
plt.ylabel("Value")
plt.grid(True)
plt.show()
```

```python
import numpy as np
from sklearn.ensemble import IsolationForest
from sklearn.metrics import confusion_matrix, classification_report
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import seaborn as sns

# === 1️⃣ Prepare Data ===
X_train = train_df.drop(columns=['id', 'Activity'])
y_train = np.where(train_df['Activity'] == 0, -1, 1)   # -1 = anomaly, 1 = normal

X_test = test_df.drop(columns=['id', 'Activity'])
y_test = np.where(test_df['Activity'] == 0, -1, 1)

print("Train shape:", X_train.shape)
print("Test shape :", X_test.shape)

# === 2️⃣ Train Isolation Forest ===
iso = IsolationForest(
    n_estimators=200,
    contamination=0.05,   # approximate anomaly fraction
    random_state=42,
    n_jobs=-1
)
iso.fit(X_train)

# === 3️⃣ Predict on Test Data ===
y_pred = iso.predict(X_test)         # -1 anomaly, 1 normal
scores = iso.decision_function(X_test)  # anomaly score (higher = more normal)

# === 4️⃣ Evaluate Accuracy ===
print("\n📌 Confusion Matrix (Labels: Normal=1, Anomaly=-1):")
cm = confusion_matrix(y_test, y_pred, labels=[1, -1])
print(cm)

print("\n📊 Classification Report:")
print(classification_report(y_test, y_pred, target_names=["Normal", "Anomaly"]))

# === 5️⃣ Visualize Anomaly Score Distribution ===
plt.figure(figsize=(10,5))
sns.histplot(scores[y_test == 1], color='green', label='True Normal', bins=30, kde=True)
sns.histplot(scores[y_test == -1], color='red', label='True Anomaly', bins=30, kde=True)
plt.axvline(0, color='black', linestyle='--')
plt.title("Isolation Forest Anomaly Score Distribution")
plt.xlabel("Anomaly Score (higher = more normal)")
plt.ylabel("Count")
plt.legend()
plt.show()

# === 6️⃣ Visualize Anomalies in 2D using PCA ===
pca = PCA(n_components=2)
X_test_pca = pca.fit_transform(X_test)

plt.figure(figsize=(8,6))
plt.scatter(X_test_pca[y_pred == 1, 0], X_test_pca[y_pred == 1, 1],
            c='green', label='Pred Normal', alpha=0.5, s=20)
plt.scatter(X_test_pca[y_pred == -1, 0], X_test_pca[y_pred == -1, 1],
            c='red', label='Pred Anomaly', alpha=0.6, s=20)
plt.title("Isolation Forest Predicted Anomalies (PCA 2D Projection)")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.legend()
plt.show()
```

```python
total_samples = len(y_pred)
anomalies_detected = np.sum(y_pred == -1)
normal_detected = np.sum(y_pred == 1)

print(f"\nTotal samples in test set: {total_samples}")
print(f"Number of anomalies detected: {anomalies_detected}")
print(f"Number of normal points detected: {normal_detected}")
```

```python
from sklearn.metrics import accuracy_score

# y_test: true labels (-1 = anomaly, 1 = normal)
# y_pred: predicted labels (-1 = anomaly, 1 = normal)
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy of Isolation Forest on test set: {accuracy*100:.2f}%")
```
